```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/kseb_electricity_consumption.csv.csv")
print(df)
```

```python
df["Percentage"] = (df["Consumption"] / df["Consumption"].sum()) * 100
df
```

```python

plt.figure(figsize=(8,5))
plt.bar(df["Category"], df["Consumption"])
plt.xticks(rotation=45)
plt.title("Kerala Electricity Consumption by Category")
plt.xlabel("Category")
plt.ylabel("Consumption")
plt.show()
```

```python
import matplotlib.pyplot as plt
colors = ["#ff9999", "#66b3ff", "#99ff99", "#ffcc99", "#c2c2f0", "#ffb3e6"]
plt.figure(figsize=(6,6))
plt.pie(df["Consumption"], labels=df["Category"], autopct="%1.1f%%", colors=colors)
plt.title("Consumption Distribution")
plt.show()
```

**KEY INSIGHTS**


* Domestic sector consumes the highest electricity

* Commercial sector is the second highest contributor


* Industrial and agriculture sectors contribute comparatively less
*   Public lighting has the lowest electricity consumption
